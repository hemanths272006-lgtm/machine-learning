```python
irisIndData=pd.DataFrame(irisdata.data)
irisIndData.head(5)
```

```python
irisdata.feature_names
```

```python
irisIndData.columns = ['sepal_length','sepal_width','petal_length','petal_width']
irisIndData.head(5)
```

```python
irisIndTrain,irisIndTest,irisDepTrain,irisDepTest = train_test_split(irisIndData,irisdata.target,test_size=0.25)
```

```python
dt = tree.DecisionTreeClassifier(max_depth=4,criterion='entropy')
```

```python
modl = dt.fit(irisIndTrain,irisDepTrain)
```

```python
predictedData=modl.predict(irisIndTest)
predictedData
```

```python
import graphviz
```

```python
dot_data = tree.export_graphviz(modl,out_file=None,feature_names=irisIndData.columns,class_names=irisdata.target_names,filled=True,rounded=True,special_characters=True)
```

```python
graph=graphviz.Source(dot_data)
graph
```
